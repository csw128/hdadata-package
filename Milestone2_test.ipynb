## Biological Use Case Example

#### Use Case: Identify dynamically regulated transcriptional programs during cortisol stress response using highly variable NTR genes

| Analysis | Biological Interpretation |
| --------- | ----------- | 
| Gene-gene similarity | co-regulated dynamic stress-response genes |
| Cell-cell similarity | cells with similar transcriptional stress states |
| Trajectory similarity | similar temporal response paths |

#### 1. Highly Variable Genes (HVGs)

To obtain a meaning subset from the sparse matrix to analyze, first, we need to identify the Highly Variable Genes (HVG). HVGs are genes which are actively changing during stress response. HVGs are identified using NTR values. The NTR variance for each gene across cells is computed to identify genes with the highest variances, which indicate cells that are active, such as initiating stress response.

```python
top_10_genes = scifate.get_highly_variable_genes(n=10)
print(f'Top 10 gene names: {top_10_genes}')
```

```python
# top_10_genes  = ['YOD1', 'SGK1', 'PARD6B', 'SMAD7', 'KLF6', 'KLF10', 'CEBPD', 'PLK2', 'ADM', 'GRPEL2']
```

```python
subset_top10 = scifate[top_10_genes, :]
subset_top10
```

```python
# # top10_subset.layers["ntr"]
# df_test = top10_subset.show(top10_subset.layers["ntr"], n_rows=10, n_cols=5, mode="label")
# df_test
```

#### 2. Gene-gene similarity 

Gene-gene similarity with raw ntr

```python
scifate.compute_similarity(
    data=subset_top10,
    key="ntr",
    kind="gene",
    method="correlation",
    name="gg_hvg_ntr"
)
```

```python
gg_hvg_ntr = scifate.similarities["gg_hvg_ntr"]
gg_hvg_ntr
```

```python
plot_heatmap(sim=gg_hvg_ntr, type="gene", cluster=True)
```

Gene-gene similarity with trajectory-based mean ntr

```python
scifate.compute_similarity(
    data=subset_top10,
    key="mean_ntr",
    kind="gene",
    method="correlation",
    name="gg_hvg_mean_ntr"
)
```

```python
gg_hvg_mean_ntr = scifate.similarities["gg_hvg_mean_ntr"]
gg_hvg_mean_ntr 
```

```python
plot_heatmap(sim=gg_hvg_mean_ntr, type="gene", cluster=True)
```

#### 3. Cell-cell similarity

Cell-cell similarity with normalized expression values

```python
cells_hvg = scifate.get_timepoint_cells(n=2)
cells_hvg
```

```python
subset_hvg_cell = scifate[top_10_genes, cells_hvg]
subset_hvg_cell
```

```python
scifate.compute_similarity(
    data=subset_hvg_cell,
    key="X",
    kind="cell",
    method="correlation",
    name="cc_hvg_X"
)
```

```python
cc_hvg_X = scifate.similarities["cc_hvg_X"]
cc_hvg_X.head()
```

```python
plot_heatmap(sim=cc_hvg_X, type="cell", cluster=True)
```

#### 4. Trajectory-trajectory similarity

Trajectory-trajectory similarity with ntr based on HVGs

```python
top_10_traj = subset_top10.get_highly_dynamic_trajectories(n=10)
top_10_traj
```

```python
scifate.compute_similarity(
    data=scifate,
    key="ntr",
    kind="trajectory",
    trajectory_id=top_10_traj,
    method="correlation",
    name="tt_hvg_ntr"
)
```

```python
tt_hvg_ntr = scifate.similarities["tt_hvg_ntr"]
tt_hvg_ntr.head()
```

```python
plot_heatmap(sim=tt_hvg_ntr, type="trajectories", cluster=False)
```
