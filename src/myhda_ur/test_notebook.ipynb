```python
import pandas as pd
```

```python
def read_tsv(path: str) -> pd.DataFrame:
    return pd.read_csv(path, sep="\t", header=None)
```

```python
var = read_tsv("../../data/features.tsv.gz")
var
```

```python
var.index = var.iloc[:, 1]
var
```

```python
var = var.drop(columns=[var.columns[1]]) 
var
```

```python
# gene name to index map

gene_index = {name: i for i, name in enumerate(var.index)}
gene_index
```

```python
# index to gene map

index_gene = {i: name for i, name in enumerate(var.index)}
index_gene
```
