```python
from read.loader import Loader
from structure.object import HDAData
from algo.distance import compute_similarity
from visualize.plot import plot_heatmap
```

```python
loader = Loader()
data = loader.load_as_hda()
```

```python
subset = data[50:60, 50:60]
```

```python
subset.show(subset.layers["ntr"], mode="label")
```

```python
compute_similarity(
    data=subset,
    key="ntr",
    axis="gene",
    method="correlation",
    target=data,
    name="gene_gene_sim_ntr"
)
```

```python
data.similarities["gene_gene_sim_ntr"]
```

```python
compute_similarity(
    data=subset,
    key="X",
    axis="cell",
    method="correlation",
    target=data,
    name="cell_cell_X"
)
```

```python
data.similarities["cell_cell_X"].head()
```

```python
df1 = data.similarities["gene_gene_sim_ntr"]
plot_heatmap(df1, "gene")
```

```python
df3 = data.similarities["cell_cell_X"]
plot_heatmap(df3, "cell")
```
