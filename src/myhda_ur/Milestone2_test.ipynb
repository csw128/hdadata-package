```python
from load.loader import Loader
from dynamic_models.similarity import compute_similarity
from visualize.plot import plot_heatmap
from dynamic_models.selection import get_highly_variable_genes
```

```python
loader = Loader()
data = loader.load_as_hda()
```

```python
subset = data[50:60, 50:60]
```

```python
subset.show(subset.layers["ntr"], mode="label")
```

Gene-gene similarity

```python
compute_similarity(
    data=subset,
    key="ntr",
    kind="gene",
    method="correlation",
    target=data,
    name="gene_gene_sim_ntr"
)
```

```python
data.similarities["gene_gene_sim_ntr"]
```

```python
df1 = data.similarities["gene_gene_sim_ntr"]
plot_heatmap(df1, "gene")
```

Cell-cell similarity

```python
compute_similarity(
    data=subset,
    key="X",
    kind="cell",
    method="correlation",
    target=data,
    name="cell_cell_X"
)
```

```python
data.similarities["cell_cell_X"].head()
```

```python
df3 = data.similarities["cell_cell_X"]
plot_heatmap(df3, "cell")
```

Trajectory-trajectory similarity

```python
compute_similarity(
    data=data,
    key="ntr",
    kind="trajectory",
    trajectory_id=data.trajectories.index[:10],
    method="correlation",
    name="traj_traj_ntr"
)
```

```python
data.similarities["traj_traj_ntr"].head()
```

```python
df4 = data.similarities["traj_traj_ntr"]
plot_heatmap(df4, "trajectories")
```

## Biological Use Case Example

#### Use Case: Identify dynamically regulated transcriptional programs during cortisol stress response using highly variable NTR genes

| Analysis | Biological Interpretation |
| --------- | ----------- | 
| Gene-gene similarity | Which genes respond together over time? |
| Cell-cell similarity | Which cells show similar stress states? |
| Trajectory similarity | Which temporal responses follow similar dynamics? |

#### 1. Highly Variable Genes (HVGs)

To obtain a meaning subset from the sparse matrix to analyze, first, we need to identify the Highly Variable Genes (HVG). HVGs are genes which are actively changing during stress response. HVGs are identified using NTR values. The NTR variance for each gene across cells is computed to identify genes with the highest variances, which indicate cells that are active, such as initiating stress response.

```python
top_10 = get_highly_variable_genes(data=data, n=10)
```

```python
print(f'Top 10 gene names: {top_10}')
```

```python
top10_subset = data[['YOD1', 'SGK1', 'PARD6B', 'SMAD7', 'KLF6', 'KLF10', 'CEBPD', 'PLK2', 'ADM', 'GRPEL2'], :]
top10_subset
```
